```python
from scipy.stats import ttest_ind

female = df[df["Gender"] == "Female"]["Performance_Score"]
male = df[df["Gender"] == "Male"]["Performance_Score"]

t_stat, p_value = ttest_ind(female, male)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis (H0)")
    print("There is a significant difference in performance between Male and Female employees.")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print("There is no significant difference in performance between Male and Female employees.")
```

```python
from scipy.stats import f_oneway

it = df[df["Department"] == "IT"]["Performance_Score"]
marketing = df[df["Department"] == "Marketing"]["Performance_Score"]
sales = df[df["Department"] == "Sales"]["Performance_Score"]

f_stat, p_value_anova = f_oneway(it, marketing, sales)

print("F-statistic:", f_stat)
print("P-value:", p_value_anova)
```

```python
alpha = 0.05

if p_value_anova < alpha:
    print("Reject the Null Hypothesis (H0)")
    print("There is a significant difference in performance among the departments.")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print("There is no significant difference in performance among the departments.")
```

```python
from scipy.stats import shapiro

performance = df["Performance_Score"]

stat, p = shapiro(performance)

print("Shapiro-Wilk Test")
print("Statistic:", stat)
print("P-value:", p)

if p < 0.05:
    print("Result: Data is not normally distributed.")
else:
    print("Result: Data is approximately normally distributed.")
```

```python
from scipy.stats import kstest

performance = df["Performance_Score"]

stat, p = kstest(
    performance,
    "norm",
    args=(performance.mean(), performance.std())
)

print("Kolmogorov-Smirnov Test")
print("Statistic:", stat)
print("P-value:", p)

if p < 0.05:
    print("Result: Data is not normally distributed.")
else:
    print("Result: Data is approximately normally distributed.")
```

```python
from scipy.stats import kstest
import numpy as np

performance = df["Performance_Score"]

mean = performance.mean()
std = performance.std()

standardized = (performance - mean) / std

stat, p = kstest(standardized, "norm")

print("Kolmogorov-Smirnov Test")
print("Statistic:", stat)
print("P-value:", p)

if p < 0.05:
    print("Result: Data is not normally distributed.")
else:
    print("Result: Data is approximately normally distributed.")
```

```python
from scipy.stats import mannwhitneyu

female = df[df["Gender"] == "Female"]["Performance_Score"]
male = df[df["Gender"] == "Male"]["Performance_Score"]

u_stat, p_value_mw = mannwhitneyu(
    female,
    male,
    alternative="two-sided"
)

print("Mann-Whitney U Test")
print("U-statistic:", u_stat)
print("P-value:", p_value_mw)

if p_value_mw < 0.05:
    print("Result: Significant difference between Male and Female performance.")
else:
    print("Result: No significant difference between Male and Female performance.")
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

model = ols(
    "Performance_Score ~ C(Department) + C(Gender) + C(Department):C(Gender)",
    data=df
).fit()

anova_table = sm.stats.anova_lm(model, typ=2)

print("Two-Way ANOVA Results")
print(anova_table)
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey_result = pairwise_tukeyhsd(
    endog=df["Performance_Score"],
    groups=df["Department"],
    alpha=0.05
)

print("Tukey HSD Post-Hoc Test")
print(tukey_result)
```

## Hypotheses and Confidence Interval

### 1. Independent Samples t-Test

**Null Hypothesis (H₀):** There is no significant difference in mean performance scores between male and female employees.

**Alternative Hypothesis (H₁):** There is a significant difference in mean performance scores between male and female employees.

**Significance Level:** α = 0.05

**Confidence Level:** 95%

---

### 2. Mann-Whitney U Test

**Null Hypothesis (H₀):** The performance score distributions of male and female employees are the same.

**Alternative Hypothesis (H₁):** The performance score distributions of male and female employees are different.

**Significance Level:** α = 0.05

**Confidence Level:** 95%

---

### 3. One-Way ANOVA

**Null Hypothesis (H₀):** The mean performance scores are equal across IT, Marketing, and Sales departments.

**Alternative Hypothesis (H₁):** At least one department has a different mean performance score.

**Significance Level:** α = 0.05

**Confidence Level:** 95%

---

### 4. Two-Way ANOVA

**Null Hypothesis (H₀):** Department, Gender, and their interaction have no significant effect on performance score.

**Alternative Hypothesis (H₁):** At least one factor or their interaction has a significant effect on performance score.

**Significance Level:** α = 0.05

**Confidence Level:** 95%

### Overall Interpretation

A p-value less than 0.05 indicates a statistically significant result. A p-value greater than or equal to 0.05 indicates that the null hypothesis is not rejected.

## Conclusion

The employee performance dataset containing 120 employees was analyzed using descriptive statistics, normality tests, hypothesis testing, ANOVA, correlation analysis, and data visualization.

The Shapiro-Wilk and Kolmogorov-Smirnov tests were used to assess the normality of the performance score distribution. The independent samples t-test and Mann-Whitney U test were conducted to compare the performance of male and female employees. The results indicated that there was no statistically significant difference in performance between the two gender groups at the 5% significance level.

One-Way ANOVA was performed to compare performance scores across the IT, Marketing, and Sales departments. The results showed no statistically significant difference among the departments.

Two-Way ANOVA was used to examine the effects of Department and Gender, including their interaction, on performance scores. Tukey HSD post-hoc analysis was also performed to identify possible differences between department groups.

Correlation analysis showed positive relationships between Experience, Sales, and Performance Score. In particular, employee experience showed a positive association with both sales and performance.

Overall, the statistical analysis provides evidence that employee experience is positively associated with sales and performance, while the observed differences based on gender and department were not statistically significant at the 0.05 significance level.
